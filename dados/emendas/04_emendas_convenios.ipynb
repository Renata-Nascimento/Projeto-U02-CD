```python
import os, re, sys, glob, unicodedata
import numpy as np
import pandas as pd
import requests

emendas_convenio = pd.read_csv('https://raw.githubusercontent.com/Renata-Nascimento/Projeto-U02-CD/main/dados/emendas/EmendasParlamentares_Convenios.csv', sep=';', encoding='latin1')
```

```python
# Filtrando os dados onde a localidade contém 'RN'
emendas_rn = emendas_convenio[emendas_convenio['Localidade do gasto'].str.contains('RN', na=False, case=False)]

# Exibindo as primeiras linhas do resultado
emendas_rn.head()
```

```python
# Converter a coluna de data para o formato datetime
emendas_convenio['Data Publicação Convênio'] = pd.to_datetime(emendas_convenio['Data Publicação Convênio'], format='%d/%m/%Y', errors='coerce')

# Filtrar os dados para manter apenas os anos de 2021 em diante
emendas_2021_mais = emendas_convenio[emendas_convenio['Data Publicação Convênio'].dt.year >= 2021]

# Exibir o resultado
emendas_2021_mais.head()
```

```python
# Garantir que a coluna de data está no formato datetime
#emendas_convenio['Data Publicação Convênio'] = pd.to_datetime(emendas_convenio['Data Publicação Convênio'], format='%d/%m/%Y', errors='coerce')

# Filtrar localidade (RN) E ano (>= 2021)
filtro_rn_2021 = (
    emendas_convenio['Localidade do gasto'].str.contains('RN', na=False, case=False) &
    (emendas_convenio['Data Publicação Convênio'].dt.year >= 2021)
)

emendas_rn_2021 = emendas_convenio[filtro_rn_2021]

# Ordenar os resultados pela data de publicação (da mais recente para a mais antiga)
emendas_rn_2021_ordenado = emendas_rn_2021.sort_values(by='Data Publicação Convênio', ascending=False)

# Exibir os dados ordenados
emendas_rn_2021_ordenado.head()
```

```python
emendas_rn_2021_ordenado.info()
```

```python
# Filtrar apenas o estado do RN
emendas_rn_filtrado = emendas_convenio[emendas_convenio['Localidade do gasto'].str.contains('RN', na=False, case=False)]

# Ordenar alfabeticamente pela coluna 'Localidade do gasto' (Municípios)
emendas_rn_ordenado_municipio = emendas_rn_filtrado.sort_values(by='Localidade do gasto')

# Exibir os dados ordenados
emendas_rn_ordenado_municipio.head()
```

```python
emendas_rn_ordenado_municipio.info()
```

```python
# Filtrar garantindo que a localidade termina com '- RN'
filtro_rn_estrito = (
    emendas_convenio['Localidade do gasto'].str.endswith('- RN', na=False) |
    (emendas_convenio['Localidade do gasto'] == 'RIO GRANDE DO NORTE (UF)')
)
emendas_rn_estrito = emendas_convenio[filtro_rn_estrito]

# Obter e ordenar a lista de municípios únicos contemplados de fato no RN
municipios_rn = emendas_rn_estrito['Localidade do gasto'].unique()
municipios_rn_ordenados = sorted(municipios_rn)

print(f"Total correto de localidades/municípios contemplados no RN: {len(municipios_rn_ordenados)}\n")
for municipio in municipios_rn_ordenados:
    print(municipio)
```
