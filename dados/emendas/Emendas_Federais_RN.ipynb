```python


# Agrupando por Função e somando os valores em ordem decrescente
recursos_por_funcao_plot = df_analise.groupby('Função')['Vl. Emenda'].sum().sort_values(ascending=False).reset_index()

# Plotando o gráfico de barras
plt.figure(figsize=(12, 7))
sns.barplot(data=recursos_por_funcao_plot, x='Vl. Emenda', y='Função', palette='magma', hue='Função', legend=False)
plt.title('Total de Recursos Destinados por Função (em R$)', fontsize=14, fontweight='bold')
plt.xlabel('Valor Total da Emenda (R$)', fontsize=12)
plt.ylabel('Função', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
