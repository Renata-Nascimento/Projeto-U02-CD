```python
# Fonte: https://portaldatransparencia.gov.br/convenios/visao-geral

import pandas as pd

convenio = pd.read_csv('/content/convenio.csv', sep=';')

convenio.info()
```

```python
convenio.head()
```

```python
# Converter diretamente as colunas de data e valor
convenio['Ano'] = pd.to_datetime(convenio['Data de início da vigência'], format='%d/%m/%Y', errors='coerce').dt.year

convenio['Valor Recebido'] = convenio['Valor Liberado (R$)'].astype(str) \
    .str.replace('.', '', regex=False) \
    .str.replace(',', '.', regex=False) \
    .astype(float)

# Selecionar e ordenar as colunas
resultado = convenio[['Município', 'Ano', 'Valor Recebido']]

# Exibir
display(resultado.head(20))
```

```python
# Filtrar apenas o município de NATAL
natal_df = convenio[convenio['Município'].str.upper() == 'NATAL']

# Agrupar por Ano e somar o Valor Recebido
natal_consolidado = natal_df.groupby('Ano', as_index=False)['Valor Recebido'].sum()

# Adicionar uma coluna com o nome do Município para clareza
natal_consolidado.insert(0, 'Município', 'NATAL')

# Formatar a exibição do valor recebido em formato de moeda
natal_exibicao = natal_consolidado.copy()
natal_exibicao['Valor Recebido (R$)'] = natal_exibicao['Valor Recebido'].map('R$ {:,.2f}'.format)

# Exibir o resultado
display(natal_exibicao[['Município', 'Ano', 'Valor Recebido (R$)']])
```

```python
# Converter a data para o formato datetime
convenio['Data de início da vigência'] = pd.to_datetime(
    convenio['Data de início da vigência'],
    format='%d/%m/%Y',
    errors='coerce'
)

# Converter o valor liberado para formato numérico
#(garantindo conversão para string antes do replace)
convenio['Valor Liberado (R$)'] = (
    convenio['Valor Liberado (R$)']
    .astype(str)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)

convenio['Valor Liberado (R$)'] = pd.to_numeric(
    convenio['Valor Liberado (R$)'],
    errors='coerce'
)

# Extrair o ano
convenio['Ano'] = convenio['Data de início da vigência'].dt.year

# Agrupar por município e ano, somando os valores liberados
resultado = (
    convenio.groupby(['Município', 'Ano'])['Valor Liberado (R$)']
    .sum()
    .reset_index()
)

# Ordenar por município e ano
resultado = resultado.sort_values(['Município', 'Ano'])

print(resultado.head(100))
```

```python
# Criar uma cópia do resultado para aplicar a formatação visual
resultado_formatado = resultado.copy()

# Formatar a coluna de valor para o padrão de moeda brasileiro R$
resultado_formatado['Valor Liberado (R$)'] = resultado_formatado['Valor Liberado (R$)'].apply(
    lambda x: f"R$ {x:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.') if pd.notnull(x) else "R$ 0,00"
)

display(resultado_formatado.head(30))
```

Uma observação importante

Exemplo, temos:

Data de início da vigência: 28/12/1995

Data de publicação: 02/01/1996

Valor liberado: R$ 12.600,00

No código acima, o valor será contabilizado em 1995, pois estamos utilizando o ano de início da vigência.

```python
# Filtrar o DataFrame formatado para exibir apenas o ano de 2026
resultado_2026 = resultado_formatado[resultado_formatado['Ano'] == 2026]

# Exibir o resultado
display(resultado_2026)
```
